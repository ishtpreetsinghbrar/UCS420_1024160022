Q6: Employee dataset and analysis

```python
# employees.csv is included with this notebook.
employees = pd.read_csv('employees.csv', parse_dates=['Joining_Date'])
employees
```

```python
# a) Shape
print('Shape (rows, columns):', employees.shape)

# b) Data types and non-null counts
employees.info()

# c) Descriptive statistics
employees.describe(include='all')
```

```python
# d) First five and last three rows
print('First 5 rows:')
display(employees.head())
print('Last 3 rows:')
display(employees.tail(3))
```

```python
# e) Required statistics
print('Average salary:', employees['Salary'].mean())
print('Total bonus:', employees['Bonus'].sum())
print('Youngest employee age:', employees['Age'].min())
print('Highest performance rating:', employees['Rating'].max())
```

```python
# f) Sort salaries from highest to lowest
employees_sorted = employees.sort_values(by='Salary', ascending=False)
employees_sorted
```

```python
# g) Rating category
employees['Performance_Category'] = np.select(
    [employees['Rating'] >= 4.5, employees['Rating'] >= 4.0],
    ['Excellent', 'Good'], default='Average'
)
employees[['Name', 'Rating', 'Performance_Category']]
```

```python
# h) Missing values
employees.isna().sum()
```

```python
# i) Rename Employee_ID to ID
employees = employees.rename(columns={'Employee_ID': 'ID'})
employees.head()
```

```python
# j) Filter employees
more_than_five_years = employees[employees['Years_of_Experience'] > 5]
it_employees = employees[employees['Department'] == 'IT']
print('More than 5 years of experience:')
display(more_than_five_years)
print('IT department:')
display(it_employees)
```

```python
# k) Tax is 10% of salary
employees['Tax'] = employees['Salary'] * 0.10

# l) Save the modified DataFrame
employees.to_csv('employees_modified.csv', index=False)
employees
```
