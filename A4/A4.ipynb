```python
import numpy as np
array = np.array([1, 2, 3, 4, 5])
print("Numpy Array:", array)

# Addition of 2 in all the elements of the array
print("Array after adding 2:", array + 2)

# Multiplication of 3 in all the elements of the array
print("Array after multiplying by 3:", array * 3)

# Division of all the elements of the array by 2
print("Array after dividing by 2:", array / 2)
```

```python
# part a 
arr = np.array([1,2,3,6,4,5])
reverse_arr = arr[::-1]
print("Original Array:", arr)
print("Reversed Array:", reverse_arr)

# part b 
x = np.array([1,2,3,4,5,1,2,1,1,1])
values , counts = np.unique(x, return_counts=True)
most_frequent_value = values[np.argmax(counts)]
indices = np.where(x == most_frequent_value)[0]
print("Most Frequent Value of x :", most_frequent_value)
print("Indices of Most Frequent Value of x :", indices)

y = np.array([1,1,1,2,3,4,2,4,3,3])
values , counts = np.unique(y, return_counts=True)
most_frequent_value = values[np.argmax(counts)]
indices = np.where(y == most_frequent_value)[0]
print("Most Frequent Value of y :", most_frequent_value)
print("Indices of Most Frequent Value of y :", indices)
```

```python
import numpy as np 
Ishtpreet = np.linspace(10,100,25)
print("Array:" ,Ishtpreet)
print("dimensions of the array:",Ishtpreet.ndim)
print("Shape of the array:",Ishtpreet.shape)
print("Total elements of the array:",Ishtpreet.size )
print("Data type of the arry ",Ishtpreet.dtype)
print("Total no. of byte consumed :",Ishtpreet.nbytes)

#transpose of the array using reshape()
Transpose_arr = Ishtpreet.reshape(25,)
print("Transpose using reshape():", Transpose_arr)

#transpose using t attribute 
print("Transpose using t attribute:", Ishtpreet.T)
```

```python
import numpy as np
ucs420_Ishtpreet = np.array([[10,20,30,40],[50,60,70,80],[90,15,20,35]])
print("Original Array:\n", ucs420_Ishtpreet)

print("Mean of the array:", np.mean(ucs420_Ishtpreet))
print ("Median of the array:", np.median(ucs420_Ishtpreet))
print("Max of the array",np.max(ucs420_Ishtpreet))
print("Min of the array",np.min(ucs420_Ishtpreet))
print("unique elements of the array ", np.unique(ucs420_Ishtpreet))

# reshaping the array to 4x3
reshaped_ucs420_Ishtpreet = ucs420_Ishtpreet.reshape(4,3)
print("Resaped Array ", reshaped_ucs420_Ishtpreet)

# resizing the array to 2x3
resized_ucs420_Ishtpreet = np.resize(ucs420_Ishtpreet, (2,3))
print("Resized Array ", resized_ucs420_Ishtpreet)
```
