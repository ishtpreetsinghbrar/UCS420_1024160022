Assingment 5.3: WAP to add all prime numbers from 1 to n and n is given by the user.

```python
n= int(input(""))
sum = 0
print("n = ",n)

for num in range(1,n):
    if num <= 2:
        if num == 2:
            sum=sum+num
            continue
        continue
    if num%2==0:
        continue
    
    for j in range(3, int(num**(0.5)) + 1,2):
        if num % j == 0:
            continue
    sum=sum+num
print (sum)
```

Assingment 6.1: WAP using function that add all odd numbers from 1 to n, n is given by the user.

```python
n= int(input(""))
sum = 0
print("n = ",n)

def iseven(num):
    if num%2==0:
        return False

    return True


for i in range(1,n+1):
    if iseven(i):
        sum= sum + i
print (sum)
```

Assingment 6.2: WAP using function that add all prime numbers from 1 to n, n given by the user.

```python
n= int(input(""))
sum = 0
print("n = ",n)

def isprime(num):
    if num <= 2:
        if num == 2:
            return True
        return False
    if num%2==0:
        return False

    for j in range(3, int(num**(0.5)) + 1,2):
        if num % j == 0:
            return False

    return True


for i in range(1,n):
    if isprime(i):
        sum= sum + i
print (sum)
```
